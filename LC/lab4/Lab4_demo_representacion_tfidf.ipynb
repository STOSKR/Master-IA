<h1 align="center"> Lingüística Computacional - MIARFID</h1>
<h1 align="center">Extracción de características, de texto a vectores dispersos</h1>


```python
# Funciones auxiliares

import json
import pandas as pd
import re


def tokenizador_simple(f):
    tr = re.compile(r'\w+')
    return tr.findall(f)

def tokenizador_simple2(f):
    tr = re.compile(r'\W+')
    return tr.sub(' ', f).split()

def tokenizador_tonto(f):
    return f.split()

pd.set_option('display.max_columns', None)

def pinta_matriz_dispersa(M, nombre_col, pre=3, titulo=None):    
    # Crear el DataFrame
    df = pd.DataFrame(M.toarray(), columns=nombre_col)
    df = df.style.set_properties(**{'text-align': 'center'}).\
        format(lambda val: '-' if val == 0 else f'{val:.{pre}f}')
    if titulo is not None:
        df = df.set_caption(titulo)
    display(df)
```

```python
# Fragmento de un poema de Juan Ramón Jiménez
docs = ["¡Rosa no presentida, que quitara", "a la rosa la rosa, que le diera", "a la rosa le diera la rosa!"]
print(json.dumps(docs, indent=2, ensure_ascii=False))
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vec = CountVectorizer()
X = vec.fit_transform(docs)
voca = list(vec.get_feature_names_out())

print(json.dumps(docs, indent=1, ensure_ascii=False))
pinta_matriz_dispersa(X, voca, 0)
```

# sklearn.feature_extraction.text.TfidfVectorizer

https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.TfidfVectorizer.html

Convierte una colección de documentos de texto en una matriz de coheficiente TF-IDF.

Equivalente a utilizar CountVectorizer seguido de TfidfTransformer.

class sklearn.feature_extraction.text.TfidfVectorizer(input=u'content', encoding=u'utf-8', decode_error=u'strict', strip_accents=None, lowercase=True, preprocessor=None, tokenizer=None, analyzer=u'word', stop_words=None, token_pattern=u'(?u)\b\w\w+\b', ngram_range=(1, 1), max_df=1.0, min_df=1, max_features=None, vocabulary=None, binary=False, dtype=<type 'numpy.int64'>, norm=u'l2', use_idf=True, smooth_idf=True, sublinear_tf=False)[source]

## tf-idf (Term frequency – Inverse document frequency)

$$
\color{Blue}{
tf(t,d) = \text{número de veces que el token $t$ está en el document $d$}}
$$


$$
\color{Red}{
idf(t,D) = \left\{\begin{matrix} 
{\text{no suavizado} : ln \left ( \frac{|D|}{|\{d\in D: t \in d\}|} \right ) }
\\ 
{\text{suavizado}  : ln \left ( \frac{|D|+1}{ | \{d\in D: t \in d\} |+1} \right )}
\end{matrix}\right.}
$$

$$ tf\text{-}idf = \color{Blue}{tf} \cdot (\color{Red}{idf} + 1) $$



```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

#docs = ["Rosa no presentida que quitara", "a la rosa la rosa que le diera", "a la rosa le diera la rosa."]
print(json.dumps(docs, indent=1, ensure_ascii=False))

titulo = '\nContadores binarios'
vec = CountVectorizer(binary=True)
X = vec.fit_transform(docs)
voca = list(vec.get_feature_names_out())
pinta_matriz_dispersa(X, voca, 0, titulo=titulo)

titulo = '\nTF (equivalente a Contadores)'
vec = TfidfVectorizer(norm=None, use_idf=False)
#Equivalente a vec = CountVectorizer()
X = vec.fit_transform(docs)
voca = list(vec.get_feature_names_out())
pinta_matriz_dispersa(X, voca, 0, titulo=titulo)

titulo = "\n\nTF normalizado 'l1' (Contadores normalizados para sumar 1)"
vec = TfidfVectorizer(norm='l1', use_idf=False)
X = vec.fit_transform(docs)
voca = list(vec.get_feature_names_out())
pinta_matriz_dispersa(X, voca, titulo=titulo)


titulo = "\n\nTF normalizado 'l2' (Contadores normalizados para que el módulo del vector sea 1)"
vec = TfidfVectorizer(norm='l2', use_idf=False)
X = vec.fit_transform(docs)
voca = list(vec.get_feature_names_out())
pinta_matriz_dispersa(X, voca, titulo=titulo)


titulo = "\n\nTF-IDF"
vec = TfidfVectorizer(norm=None, smooth_idf=False)
X = vec.fit_transform(docs)
voca = list(vec.get_feature_names_out())
pinta_matriz_dispersa(X, voca, titulo=titulo)

titulo = "\n\nTF-IDF con idf suavizado"
vec = TfidfVectorizer(norm=None, smooth_idf=True)
X = vec.fit_transform(docs)
voca = list(vec.get_feature_names_out())
pinta_matriz_dispersa(X, voca, titulo=titulo)

titulo = "\n\nTF-IDF: con idf suavizado y normalización 'l1'"
vec = TfidfVectorizer(norm='l1', smooth_idf=True)
X = vec.fit_transform(docs)
voca = list(vec.get_feature_names_out())
pinta_matriz_dispersa(X, voca, titulo=titulo)

titulo = "\n\nTF-IDF: con idf suavizado y normalización 'l2'(opción por defecto)"
vec = TfidfVectorizer()
X = vec.fit_transform(docs)
voca = list(vec.get_feature_names_out())
pinta_matriz_dispersa(X, voca, titulo=titulo)
```
